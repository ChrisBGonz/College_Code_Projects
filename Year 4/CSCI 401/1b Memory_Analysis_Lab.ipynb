Run the windows.info plugin to show OS & kernel details of the memory sample being analyzed

***Note the command syntax, where the [image file name] and [plugin] are replaced with the actual names of the memory dump and the plugin used. You'll use this syntax for the rest of the commands.***

!python volatility3/vol.py -f [image file name] [plugin]

```python
!python volatility3/vol.py -f memdump.mem windows.info
```

Run the windows.pslist plugin to list the processes present

```python
!python volatility3/vol.py -f memdump.mem windows.pslist
```

Run the windows.pstree plugin to list processes in a tree based on their parent process ID

```python
!python volatility3/vol.py -f memdump.mem windows.pstree
```

Run the windows.handles plugin to find the handles associated with a process ID that you find in the process list output above.

Add --pid [pid] at the end of the plugin command. Replace [pid] with the process number you choose to use.

```python
!python volatility3/vol.py -f memdump.mem windows.handles --pid 492
```
